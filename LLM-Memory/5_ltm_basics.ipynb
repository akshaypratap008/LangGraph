```python
from langgraph.store.memory import InMemoryStore
```

```python
# creating a store
store= InMemoryStore()
```

```python
# creating a namespace      - namespaces are like folders to organise the memories
namespace = ("user", "u1")
```

Creating memories

```python
store.put(namespace=namespace, key="1", value = {"data": "User likes pizza"})
store.put(namespace=namespace, key="2", value={"data": "User prefers darkmode"})
```

```python
# create another namespace
namespace2 = ("user", "u2")

# add memories to the 2nd namespace
```

```python
store.put(namespace2, "1", {"data": "User likes pasta"})
store.put(namespace2, "2", {"data": "User likes blue color"})
```

Retrieving Memories

```python
store.get(namespace2, "1")
```

```python
# fetching all the memory - search
items = store.search(namespace2)     # for first namespace
[item.value for item in items]
```

Searching semantically for memories

```python
# import an embedding model 
from langchain_openai import OpenAIEmbeddings
from dotenv import load_dotenv
load_dotenv()

embedding_model = OpenAIEmbeddings(model = "text-embedding-3-small")
```

```python
store = InMemoryStore(index = {'embed': embedding_model, "dims": 1536})
```

```python
# define namespace
namespace = ("users", "u1")
```

```python
# add memories
store.put(namespace, "1", {"data": "user prefers consise answers over long explainations"})
store.put(namespace, "2", {"data": "user likes examples in python"})
store.put(namespace, "3", {"data": "user usually works late at night"})
store.put(namespace, "4", {"data": "user prefers dark mode applications"})
store.put(namespace, "5", {"data": "user is based in UK"})
store.put(namespace, "6", {"data": "user works in lloyds bank"})
store.put(namespace, "7", {"data": "user prefers bullet points"})
```

```python
items = store.search(namespace, query = "What is the user currently learning", limit = 1)

for item in items:
    print(item.value)
```

```python
items = store.search(namespace, query = "What is the user's preferences", limit = 3)

for item in items:
    print(item.value)
```
