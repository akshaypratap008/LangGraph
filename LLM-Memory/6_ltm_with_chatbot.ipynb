```python
from langgraph.store.memory import InMemoryStore
from langchain_openai import OpenAIEmbeddings
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
load_dotenv()

from langgraph.graph import StateGraph, MessagesState, START, END
from langchain_core.runnables import RunnableConfig
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
from langgraph.store.base import BaseStore
```

```python
store = InMemoryStore()

user_id = "u1"

# stor user details as a simple blob
user_details = ("user", user_id, "details")

store.put(user_details, "profile_1", {"data": "Name: Akshay"})
store.put(user_details, "profile_2", {"data": "Profession: Data Scientist"})
store.put(user_details, "preference_1", {"data": "prefers concise answers"})
store.put(user_details, "preference_2", {"data": "knows python"})
store.put(user_details, "project_1", {"data": "learning agentic AI right now"})
```

```python
# define a system prompt which will instruct the llm to use the memories

SYSTEM_PROMPT_TEMPLATE = """
You are a helpfull assistant with memory capabilities.
If user-specific memory is available, use it to personalize your response based on what you know about the user. 

Your goal is to provide ralevant, friendly and tailored assistance that reflects the user's preferences, context and past interactions. 

If the user's name or relevant personal context is available, always personalise your responses by:
    - Always address user by name (e.g., "Sure, Akshay..") when appropriate
    - Referencing known projects, tools, or preferences (e.g., "your agentic AI project...")
    - Adjust the tone to feel friendly, natural and directly aimed at the user. 

Avoid generic phrasing when personalisation is possible, For example, instead of "In TypeScript app..." say "Since your project is built on TypeScript"ArithmeticError

User personalisation especially in: 
    - grettings and transitions
    - Help or guidance tailored to tools and frameworks the user uses
    - Follow-up messages that continue from past context

Always ensure that personalisation is based only on known details and not assumed

In the end suggest 3 relevant futher questions based on current response and user profile.

The user's memory (which may be empty) us provided as : {user_details_content}
"""
```

```python
llm = ChatOpenAI(model= "gpt-4o-mini")
```

```python
def chat_node(state: MessagesState, config: RunnableConfig, store: BaseStore):
    # fetch user if from config dict
    user_id = config['configurable']['user_id']

    # fetch memories
    user_details = ("user", user_id, "details")
    items = store.search(user_details)

    # convert memory items into string blob to feed to user_details_content
    if items:
        user_details_content = "\n".join(item.value.get("data", "") for item in items)
    else:
        user_details_content = ""   # prompt says it may be empty

    system_prompt = SYSTEM_PROMPT_TEMPLATE.format(
        user_details_content = user_details_content
    )

    # convert system prompt into system message
    sys_msg = SystemMessage(content = system_prompt)

    # call llm
    response = llm.invoke([sys_msg] + state['messages'])

    return {"messages": [response]}
```

```python
# build graph
builder = StateGraph(MessagesState)

builder.add_node("chat", chat_node)
builder.add_edge(START, "chat")

graph = builder.compile(store = store)

graph
```

```python
config = {"configurable": {"user_id": "u1"}}

result = graph.invoke(
    {"messages": [HumanMessage(content = "What am my preferences?")]},
    config = config
)

print(result['messages'][-1].content)
```

### Creating new memories

```python
from pydantic import BaseModel, Field
from typing import List
import uuid
```

```python
# creating new store
store = InMemoryStore()
```

```python
# llm which decides what to remember
extractor_llm = ChatOpenAI(model = "gpt-4o-mini")
```

```python
class MemoryDecision(BaseModel):
    should_write:bool = Field(description="Whether to store any particular memory or not")
    memories: List[str] = Field(default_factory=list, description="Atomic user memory to store")
```

```python
memory_extractor_llm = extractor_llm.with_structured_output(MemoryDecision)
```

```python
def remember_only_node(state: MessagesState, config: RunnableConfig, store: BaseStore):
    user_id = config['configurable']['user_id']

    namespace = ("user", user_id, "details")

    # take last user message to check if anything needs storing
    last_message = state['messages'][-1]

    # send to memory extractor 
    decision:MemoryDecision = memory_extractor_llm.invoke(
        [
            SystemMessage(content = """
Extract LONG TERM MEMORY from the user's message.
Only store stable, user-specific info (identity, preferences, ongoing projects).
Do not store transient info.
Return should_write = Fasle if nothing is worth storing.
Each memory should be a short atomic sentence.
"""),
last_message
        ]
    )

    # Write to store (LTM)
    if decision.should_write:
        for mem in decision.memories:
            store.put(namespace, str(uuid.uuid4()), {"data": mem})

    # IMPORTANT: we are NOT using memory, not even responding with the LLM.
    # We just return a fixed acknowledgement.
    return {"messages": [AIMessage(content = "Noted")]}


```

```python
# build graph 
builder = StateGraph(MessagesState)

builder.add_node("remember", remember_only_node)
builder.add_edge(START, "remember")
builder.add_edge("remember", END)

graph = builder.compile(store = store)
graph
```

### Fixing Duplication issue with memory

```python
# creating a new class MemoryItem
class MemoryItem(BaseModel):
    """This will be the actual memory item. Which later will be checked if it exists in memories list or not"""
    text: str = Field(description="Atomic user memory as a short sentence")
    is_new: bool = Field(description="True if this memory is NEW and should be stored. False if duplicate/already exists or known")

class MemoryDecision(BaseModel):
    """This will contain the memories list and decision if the memory should be stored or not"""
    should_write:bool = Field(description="Whether to store any particular memory or not")
    memories: List[MemoryItem] = Field(default_factory=list, description="Atomic user memory to store")     #this will be the list of MemoryItems (memories)
```

```python
memory_extractor_llm = extractor_llm.with_structured_output(MemoryDecision)
```

```python
MEMORY_PROMPT = """
You are responsible for updating and maintaining accurate user memory. 

CURRENT USER DETAILS (existing memory): {user_details_content}

TAST:
- Review the user's latest message.
- Extract user specific info worth storing long-term(identity, stable preferences, ongoing projects/goals).
- For each extracted item, set is_new = True ONLY if it adds new information in the CURRENT USER DETAILS.
- If it contains the same meaning as something already present, set is_new = False
- Keep each memory as a short atomic sentence.
- No speculation, only facts stated by the user
- If there is nothing memory worthy, return an empty list
"""
```

```python
def chat_create_memory_node(state: MessagesState, config: RunnableConfig, store: BaseStore):
    user_id = config['configurable']['user_id']

    namespace = ("user", user_id, "details")

    # load existing memories
    existing_items = store.search(namespace)
    existing_texts = [it.value.get("data", "") for it in existing_items if it.value.get("data")]
    user_details_content = "\n".join(f" - {t}" for t in existing_texts) if existing_texts else "(empty)"

    # take last user message to check if anything needs storing
    last_message = state['messages'][-1]

    # send to memory extractor 
    decision:MemoryDecision = memory_extractor_llm.invoke(
        [SystemMessage(content = MEMORY_PROMPT.format(user_details_content = user_details_content)), last_message]
    )

    # Write to store (LTM)
    if decision.should_write:
        for mem in decision.memories:
            if mem.is_new:
                store.put(namespace, str(uuid.uuid4()), {"data": mem.text})

    # IMPORTANT: we are NOT using memory, not even responding with the LLM.
    # We just return a fixed acknowledgement.
    return {"messages": [AIMessage(content = "Noted")]}
```

```python
# build graph 
builder = StateGraph(MessagesState)

builder.add_node("chat", chat_create_memory_node)
builder.add_edge(START, "chat")
builder.add_edge("chat", END)

graph = builder.compile(store = store)
graph
```

### Merged Workflow

```python
store = InMemoryStore()
```

```python
SYSTEM_PROMPT_TEMPLATE = """You are a helpful assistant with memory capabilities.
If user-specific memory is available, use it to personalize 
your responses based on what you know about the user.

Your goal is to provide relevant, friendly, and tailored 
assistance that reflects the user’s preferences, context, and past interactions.

If the user’s name or relevant personal context is available, always personalize your responses by:
    – Always Address the user by name (e.g., "Sure, Nitish...") when appropriate
    – Referencing known projects, tools, or preferences (e.g., "your MCP server python based project")
    – Adjusting the tone to feel friendly, natural, and directly aimed at the user

Avoid generic phrasing when personalization is possible.

Use personalization especially in:
    – Greetings and transitions
    – Help or guidance tailored to tools and frameworks the user uses
    – Follow-up messages that continue from past context

Always ensure that personalization is based only on known user details and not assumed.

In the end suggest 3 relevant further questions based on the current response and user profile

The user’s memory (which may be empty) is provided as: {user_details_content}
"""
```

```python
memory_llm = ChatOpenAI(model = "gpt-4o-mini")
```

```python
# 2 pydantic class same as before

class MemoryItem(BaseModel):
    """This will be the actual memory item. Which later will be checked if it exists in memories list or not"""
    text: str = Field(description="Atomic user memory as a short sentence")
    is_new: bool = Field(description="True if this memory is NEW and should be stored. False if duplicate/already exists or known")

class MemoryDecision(BaseModel):
    """This will contain the memories list and decision if the memory should be stored or not"""
    should_write:bool = Field(description="Whether to store any particular memory or not")
    memories: List[MemoryItem] = Field(default_factory=list, description="Atomic user memory to store")    
```

```python
memory_extractor = memory_llm.with_structured_output(MemoryDecision)
```

```python
MEMORY_PROMPT = """You are responsible for updating and maintaining accurate user memory.

CURRENT USER DETAILS (existing memories):
{user_details_content}

TASK:
- Review the user's latest message.
- Extract user-specific info worth storing long-term (identity, stable preferences, ongoing projects/goals).
- For each extracted item, set is_new=true ONLY if it adds NEW information compared to CURRENT USER DETAILS.
- If it is basically the same meaning as something already present, set is_new=false.
- Keep each memory as a short atomic sentence.
- No speculation; only facts stated by the user.
- If there is nothing memory-worthy, return should_write=false and an empty list.
"""
```

```python
def remember_node(state: MessagesState, config: RunnableConfig, *, store: BaseStore):
    user_id = config["configurable"]["user_id"]
    ns = ("user", user_id, "details")

    # existing memory
    items = store.search(ns)
    existing = "\n".join(it.value["data"] for it in items) if items else "(empty)"

    # last user message
    last_msg = state["messages"][-1].content

    decision: MemoryDecision = memory_extractor.invoke(
        [
            SystemMessage(content=MEMORY_PROMPT.format(user_details_content=existing)),
            {"role": "user", "content": last_msg},
        ]
    )

    if decision.should_write:
        for mem in decision.memories:
            if mem.is_new:
                store.put(ns, str(uuid.uuid4()), {"data": mem.text})

    return {}  # no message change
```

```python
chat_llm = ChatOpenAI(model="gpt-4o-mini")
```

```python
def chat_node(state: MessagesState, config: RunnableConfig, *, store: BaseStore):
    user_id = config["configurable"]["user_id"]
    ns = ("user", user_id, "details")

    items = store.search(ns)
    user_details = "\n".join(it.value["data"] for it in items) if items else ""

    system_msg = SystemMessage(
        content=SYSTEM_PROMPT_TEMPLATE.format(
            user_details_content=user_details or "(empty)"
        )
    )

    response = chat_llm.invoke([system_msg] + state["messages"])
    return {"messages": [response]}
```

```python
builder = StateGraph(MessagesState)
builder.add_node("remember", remember_node)
builder.add_node("chat", chat_node)

builder.add_edge(START, "remember")
builder.add_edge("remember", "chat")
builder.add_edge("chat", END)

graph = builder.compile(store=store)

graph
```
