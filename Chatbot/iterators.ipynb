```python
class mera_range:

    def __init__(self, start, end):
        self.start = start
        self.end = end

    def __iter__(self):
        return mera_iterator(self)

class mera_iterator:

    def __init__(self, iterable_obj):
        self.iterabel = iterable_obj

    def __iter__(self):
        return self

    def __next__(self):
        if self.iterabel.start >= self.iterabel.end:
            raise StopIteration
        current = self.iterabel.start
        self.iterabel.start += 1
        return current
```

```python
for i in mera_range(1, 10):
    print(i)
```
