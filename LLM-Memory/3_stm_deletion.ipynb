```python
from langgraph.graph import StateGraph, MessagesState, START, END
from langgraph.checkpoint.memory import InMemorySaver

from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain.messages import RemoveMessage, HumanMessage
```

```python
load_dotenv()
```

```python
model = ChatOpenAI()
```

```python
def chat(state: MessagesState):
    response = model.invoke(state['messages'])
    return {"messages": [response]}

def delete_old_messages(state: MessagesState):
    msgs = state['messages']

    if len(msgs) > 10:
        to_remove = msgs[:6]
        return {"messages": [RemoveMessage(id = m.id) for m in to_remove]}
    return {}
```

```python
builder = StateGraph(MessagesState)
builder.add_node("chat", chat)
builder.add_node("cleanup", delete_old_messages)
```

```python
builder.add_edge(START, "chat")
builder.add_edge("chat", "cleanup")
builder.add_edge("cleanup", END)
```

```python
graph = builder.compile(checkpointer=InMemorySaver())
```

```python
graph
```

```python
config = {"configurable": {"thread_id": "t1"}}
```

```python
graph.invoke({"messages": HumanMessage(content="Hi, I am Akshay.")}, config = config)
graph.invoke({"messages": HumanMessage(content="Hi, How are you?")}, config = config)
graph.invoke({"messages": HumanMessage(content="What is langchain?")}, config = config)
graph.invoke({"messages": HumanMessage(content="What is langgraph?")}, config = config)
graph.invoke({"messages": HumanMessage(content="What is my name?")}, config = config)
graph.invoke({"messages": HumanMessage(content="Tell me about me.")}, config = config)
```

```python
snap = graph.get_state(config = config)
[m.content for m in snap.values['messages']]
```
